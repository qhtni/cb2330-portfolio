```python
# My code can also be found at the end of the excersise notebook.
# But i also copied it into a seperate notebook since I was unsure.

######                   Part B                   #####


# I picked the dot product formula:
#   ---> "Reads as how much two vectors of numbers agree - and a wieghted sum."

#Instructions:

# Task 1. Show the formula as mathematics and say in a sentence what it computes.
# Task 2. Implement it as a loop or accumulator pattern, and resist the temptation to use sum() or numpy at first.
# Task 3. Implement it again in one line with numpy.
# Task 4. Show that they agree on the same input
# Task 5. Commit the notebook to cb2330-portfolio

#######################  Task 1.  ########################

# Mathematical formula:
# sum_(i=1)^N (a_i * b_i)

# The dot product calculates a weighted sum of two vectors, which shows us how much the two vectors agree with eachother.

# Formula  ####       Meaning #################   In the code ##################

# ∑                   Sum over items               total_ = 0

# i = 1...N           Visit every item once        for i in range(len(vector_a))
# a_i                 Item i in vector a           vector_a[i]
# b_i                 Item i in vector b           vector_b[i]

#######################  Task 2.  ########################


def dploop(vector_a, vector_b):
  total_ = 0
  for i in range(len(vector_a)):
    total_ += (vector_a[i] * vector_b[i])
  return total_


#######################  Task 3.  ########################

import numpy as np

def dpnumpy(vector_a, vector_b):
  return np.dot(vector_a, vector_b)


#######################  Task 4.  ########################


vector_a = [7, 9, 5]
vector_b = [3, 2, 8]

result_l = dploop(vector_a, vector_b)
result_n = dpnumpy(vector_a, vector_b)

print(f"Result using loop: {result_l}")
print(f"Result using numpy: {result_n}")
print(f"Do they agree on the same input?: {result_l == result_n}")


#######################  Task 5.  ########################

#Done!
```
